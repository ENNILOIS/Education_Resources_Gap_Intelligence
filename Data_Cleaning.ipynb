### **HANDLING MISSING VALUES**

#### **The dataset combines two sources:**

**Lagos Bureau of Statistics (2016–2020):** recorded enrolment and teacher ratios, but not classrooms or toilets.

**DNEMIS (2024/25):** recorded classrooms and toilets, but not enrolment by gender.

Most missing values exist because a source never collected that information. They are not errors.

#### **Why I Did Not Fill The Missing Values:**
**Not with Zero(0):** 0 is a real number. A toilet ratio of 0 would wrongly suggest perfect conditions and distort the rankings and the model.

**Not with the Median:** in the 2016–2020 rows, classroom and toilet columns are completely empty, so a median would be invented data, not a real estimate.

**Not with "Not Provided":** text in a numeric column turns it into text, which stops calculations, sorting, charts and modelling.

#### **Why I left them as NaN:** Python recognises NaN as "no value". It skips NaN in calculations and still allows sorting, charting and modelling.

#### **Why I split the data:** each part is used for the task it supports.
**Machine learning uses the 2016–2020 rows,** which have four years of teacher-ratio history.

**Resource-gap ranking uses the 2024/25 rows,** which include classroom and toilet ratios.

##### Each subset keeps only the columns available for it, which removes most missing values without creating false data.

```python
# To Check & Know the Columns in the Dataset

df_clean.columns
```

#### SPLITTING THE DATA

```python
# 1. ML dataset: 2016-2020 public rows, only columns that exist there

ml_cols = [
    "LGA", "Level", "Year_Start", "Learners", "Male_Learners",
    "Female_Learners", "Female_Learners_PCT", 
    "Learner_Teacher_Ratio", "Est_Teachers",
    "LTR_Gap_vs_UBE35", "Teachers_Needed_To_Meet_UBE35", 
    "Enrolment_Change_PCT_vs_Prev_Year"
]

ml_df = df_clean[df_clean["Year_Start"] <= 2019][ml_cols]
print(ml_df.isnull().sum())

# 2. Ranking dataset: 2024/25 rows, with classroom and toilet ratios

rank_cols =[
    "LGA", "Ownership", "Level", "Learners",
    "Learner_Teacher_Ratio", "Learner_Classroom_Ratio",
    "Learner_Toilet_Ratio", "Female_Teachers_PCT",
    "Special_Needs_Learners", "LTR_Gap_vs_UBE35"
]

rank_df = df_clean[df_clean["Year_Start"] == 2024][rank_cols]
print(rank_df.isnull().sum())
```

#### **CREATING THE FINAL ML TABLES**

##### After splitting, each table still contained a small number of rows with missing values. This step keeps only complete rows for each task, without filling in any values.

##### **ML TABLE:** The 2016/17 rows were removed from the ML table because enrolment change cannot be calculated for the first year; there is no previous year to compare with. Filling these values would have introduced false data. The final ML table has 180 complete rows (2017/18–2019/20).

```python
# Make the Final ML

# ML: Leave out 2016/17 (No Previous Year To Compare)
ml_final = ml_df.dropna()
print(ml_final.shape)
print(ml_final.isnull().sum())
```

```python
missing_rows = rank_df[rank_df.isnull().any(axis=1)]
print(missing_rows.groupby(["Ownership", "Level"]).size())
```

#### **Identifying the Source of Missing Values in the Ranking Table**

##### Before deciding how to handle the missing values in the 2024/25 ranking data, I checked which rows they came from. This showed whether the gaps were spread randomly across the data or concentrated in particular school types.

##### The check found that all missing values came from three groups: private schools (20 rows), Islamiyya/Quranic schools (IQS, 20 rows) and technical/vocational schools (18 rows). Public Primary, JSS and SSS rows had no missing values.


#### **CREATING THE FINAL RANKING TABLES**

##### After splitting, each table still contained a small number of rows with missing values. This step keeps only complete rows for each task, without filling in any values.

##### **Ranking Table:** All missing values in the 2024/25 data came from private schools, Islamiyya/Quranic schools (IQS) and technical/vocational schools. Public Primary, JSS and SSS rows were complete. The ranking therefore uses only these 60 rows. These are also the schools that government and NGOs support directly with teachers and facilities, which is the focus of this tool.

```python
# Ranking: Public Primary, JSS, SSS Only (Complete Data)

rank_final = rank_df[
    (rank_df["Ownership"] == "Public") &
    (rank_df["Level"] .isin(["Primary", "JSS", "SSS"]))
]

print(rank_final.shape)
print()
print(rank_final.isnull().sum())
print()
print(rank_final.isnull().sum().sum())
```

#### **Saving the Cleaned Datasets**

##### The cleaned data was saved as three separate files for two reasons. First, the data comes from two sources that recorded different information: the 2016–2020 report had four years of student and teacher data but no classrooms or toilets, while the 2024/25 report had classrooms and toilets for one year only. Second, each part of the project needs different data. Keeping them separate means every task uses only complete, real data, with no invented values.

**Lagos_Clean_Full.csv (360 rows):** the complete cleaned dataset with missing values kept as NaN, used for exploratory data analysis and the RAG assistant.

**Lagos_ML_Data.csv (180 rows):** complete rows from 2017/18 to 2019/20, used to train and test the machine learning models.

**Lagos_Ranking_Data.csv (60 rows):** complete 2024/25 rows for public Primary, JSS and SSS schools, used to rank LGAs by resource gaps.
