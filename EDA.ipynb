```python
# HOW MANY STUDENTS DOES EACH LGA HAS?

Students = rank_df.groupby("LGA")["Learners"].sum().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(14, 6))
Students.plot(kind="bar", color="grey", ax=ax)
ax.set_title("NUMBER OF STUDENTS PER LGA", fontsize=13, fontweight="bold", color="brown")
ax.set_xlabel("LGA", color="brown", fontweight="bold")
ax.set_ylabel("NUMBER OF STUDENTS", color="brown", fontweight="bold")
ax.tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```

```python
print("LARGEST LGA :", Students.index[0], 
      f'({Students.iloc[0]:,.0f} Students)')

print()

print("SMALLEST LGA :", Students.index[-1], 
     f'({Students.iloc[-1]:,.0f} Students)')

print()

print("RANGE       :", 
      f'{Students.max() - Students.min():,.0f} Students')
```

From the dataset we could analyse that in lagos state ikorodu and alimosho with badagry as the top 3 numbers of students in the LGA. Ikorodu LGA has the higest students in Lagos state of 107,588 students and Apapa LGA has the Lowest students in Lagos State of 16,554 Students. 

```python
# WHICH LGAs LACK TEACHERS?
# WHAT TO LOOKS AT - Learner_Teacher_Ratio vs standard 35(Universal Basic Education)

Teachers = rank_df.groupby("LGA")["Learners"].mean().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(10, 6))
Teachers.plot(kind="barh", color="grey", ax=ax)
ax.set_title("STUDENTS PER TEACHER BY LGA", fontsize=13, fontweight="bold", color="brown")
ax.set_xlabel("LGA", fontweight="bold", color="brown")
ax.set_ylabel("STUDENTS PER TEACHER", fontweight="bold", color="brown")
ax.axhline(35, color="red", linestyle="..", label="standard")
ax.tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```
