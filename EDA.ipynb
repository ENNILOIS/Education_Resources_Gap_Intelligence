```python
# WHICH LGAs LACK TEACHERS?
# WHAT TO LOOKS AT - Learner_Teacher_Ratio vs standard 35(Universal Basic Education)

Teachers = rank_df.groupby("LGA")["Learner_Teacher_Ratio"].mean().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(10, 6))
Teachers.plot(kind="barh", color="grey", ax=ax)
ax.set_title("STUDENTS PER TEACHER BY LGA", fontsize=13, fontweight="bold", color="brown")
ax.set_xlabel("LGA", fontweight="bold", color="brown")
ax.set_ylabel("STUDENTS PER TEACHER", fontweight="bold", color="brown")
ax.axvline(35, color="red", linestyle="--", label="standard (35)")
ax.tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```

#### **What the chart shows:**
##### Each bar shows the number of students for every one teacher (students per teacher) in each LGA. The value is the average across public Primary, JSS and SSS schools in 2024/25.

##### For example, a value of 69 means that, on average, about 69 students share one teacher in that LGA. It does not mean there are 69 more students than teachers. Every LGA has more students than teachers; the chart shows how many students each teacher has to handle.

#### **How to read it:**
##### The red dashed line marks the Universal Basic Education (UBE) standard of 35 students per teacher.

##### **Bars that go past the red line:** each teacher has more than 35 students, so the LGA is lacking teachers. The longer the bar, the bigger the shortage.
##### **Bars that stop before the red line:** each teacher has 35 students or fewer, so the LGA is not lacking teachers.

```python
# TO SHOW THE LGA LACKING & NOT LACKING TEACHERS

LGA_table = rank_df.groupby("LGA")["Learner_Teacher_Ratio"].mean().reset_index()
LGA_table = LGA_table.rename(columns={"Learner_Teacher_Ratio": "Students_Per_Teacher"})

Cols = ["LGA", "Students_Per_Teacher"]

print(f"TOP 5 LGAs MOST LACKING TEACHERS:")
print(LGA_table.sort_values("Students_Per_Teacher", ascending=False)[Cols].head(6).to_string(index=False))

print()

print(f"TOP 5 LGAs NOT LACKING TEAHERS:")
print(LGA_table.sort_values("Students_Per_Teacher")[Cols].head(6).to_string(index=False))
```

#### **Insight:**
##### 9 of the 20 LGAs are above the standard and are lacking teachers. **Ajeromi/Ifelodun** has the highest ratio, at about 69 students per teacher, nearly double the standard. It is followed by **Badagry (about 66) and Amuwo Odofin (about 56).**

##### **Kosofe (about 22), Ikeja (about 23) and Epe (about 24)** have the lowest ratios. They are well within the standard and are not lacking teachers.

##### These results show that teacher shortages are not spread evenly across Lagos. Some LGAs urgently need more teachers, while others already meet the standard. This suggests that redistributing or recruiting teachers for the most affected LGAs could reduce the gap.

```python
# WHICH SCHOOL LEVEL SUFFERS MOST?
# WHAT TO LOOKS AT - Average ratios for Primary, JSS, SSS

School_Level = rank_df.groupby('Level')[["Learner_Teacher_Ratio", "Learner_Classroom_Ratio", "Learner_Toilet_Ratio"]].mean()
School_Level = School_Level.loc[["Primary", "JSS", "SSS"]]
School_Level.columns = ["Students Per Teacher", "Students Per Classroom", "Students Per Toilet"]

fig, ax = plt.subplots(figsize=(14,6))
School_Level.plot(kind="bar", color=["grey", "brown", "skyblue"], width=0.8, ax=ax)

for containers in ax.containers:
    ax.bar_label(containers, fmt="%.1f", fontsize=10, fontweight="bold")

ax.axhline(35, color="red", linestyle="--", label="standard (35): teachers & classrooms")
ax.axhline(30, color="green", linestyle="--", label="standard (30): toilets")

ax.set_title("AVERAGE RESOURCE RATIOS BY SCHOOL LEVEL", fontsize=13, fontweight="bold", color="brown")
ax.set_xlabel("SCHOOL LEVEL", fontweight="bold", color="brown")
ax.set_ylabel("NUMBER OF STUDENTS", fontweight="bold", color="brown")
ax.tick_params(axis="x", rotation=0)
ax.legend()

plt.tight_layout()
plt.show()
```

#### **Which School Level Suffers Most?**

#### **What the chart shows:**
The chart compares the three school levels (Primary, JSS and SSS) across all 20 LGAs in 2024/25. Each level has three bars: the average students per teacher, students per classroom and students per toilet.

#### **How to read it:**
The red dashed line marks the standard of 35 students per teacher and per classroom.
The green dashed line marks the standard of 30 students per toilet.
Bars above their line show where that school level is lacking that resource.

#### **Insight:**
**Junior Secondary Schools (JSS)** suffer the most. JSS is above the standard for all three resources: about 49 students per teacher, 71 students per classroom (about double the standard) and 54 students per toilet.

**Senior Secondary Schools (SSS)** are also heavily overcrowded, with about 65 students per classroom and 51 students per toilet. However, their teacher ratio (about 31) is within the standard.

**Primary schools** are in the best position. Their teacher ratio (about 30) and classroom ratio (about 23) are within the standard, and their toilet ratio (about 30) is right at the standard.

This shows that the pressure on resources increases at secondary level, especially in classrooms. Support should be focused first on junior secondary schools.

```python
# WHICH RESOURCES IS THE BIGGEST PROBLEM OVERALL?
# WHAT TO LOOKS AT - Count of LGAs above each standard

LGA_Avg = rank_df.groupby("LGA")[["Learner_Teacher_Ratio", "Learner_Classroom_Ratio", "Learner_Toilet_Ratio"]].mean()

Lacking_Counts = pd.Series({
    "Teachers   ": (LGA_Avg["Learner_Teacher_Ratio"] > 35).sum(),
    "Classrooms ": (LGA_Avg["Learner_Classroom_Ratio"] > 35).sum(),
    "Toilets    ": (LGA_Avg["Learner_Toilet_Ratio"] > 30).sum()
    })

fig, ax = plt.subplots(figsize=(8, 5))
Lacking_Counts.plot(kind="bar", color=["grey", "brown", "skyblue"], ax=ax)

for container in ax.containers:
    ax.bar_label(container, fontsize=11, fontweight="bold")

ax.set_title("NUMBER OF LGAs LACKING EACH RESOURCE (OUT OF 20)", fontsize=13, fontweight="bold", color="brown")
ax.set_xlabel("RESOURCE", fontweight="bold", color="brown")
ax.set_ylabel("NUMBER OF LGAs", fontweight="bold", color="brown")
ax.set_ylim(0, 22)
ax.tick_params(axis="x", rotation=0)

plt.tight_layout()
plt.show()
```

#### **What the chart shows:**
Each bar shows how many of the 20 LGAs are lacking that resource, meaning their average ratio is above the national standard (35 for teachers and classrooms, 30 for toilets).

#### **How to read it:**
The taller the bar, the more LGAs are affected by that shortage. A bar of 20 means every LGA in Lagos is lacking that resource.

#### **Insight:**
**Classrooms are the biggest problem:** all 20 LGAs are lacking classrooms. Toilets come next, lacking in 19 LGAs. Teachers are lacking in 9 LGAs, fewer than half.

This shows that facilities (classrooms and toilets) are a more widespread problem than teachers in Lagos public schools. Teacher shortages are concentrated in particular LGAs, while facility shortages affect almost the whole state.

```python
# WHICH LGAs ARE MOST LACKING & LEAST LACKING OVERALL?
# WHAT TO LOOKS AT -   Top 5 and bottom 5 across all three ratios 

LGA_Avg["Overall_Gap"] = (
    LGA_Avg["Learner_Teacher_Ratio"] / 35 +
    LGA_Avg["Learner_Classroom_Ratio"] / 35 +
    LGA_Avg["Learner_Toilet_Ratio"] / 30
) / 3

LGA_Avg = LGA_Avg.round(2)

Cols = ["Learner_Teacher_Ratio", "Learner_Classroom_Ratio", "Learner_Toilet_Ratio", "Overall_Gap"]

print("TOP 5 MOST LACKING LGAs:")
print(LGA_Avg.sort_values("Overall_Gap", ascending=False)[Cols].head(5))

print()

print("TOP 5 LEAST LACKING LGAs:")
print(LGA_Avg.sort_values("Overall_Gap", ascending=True)[Cols].head(5))
```

```python
# THE CHART

Overall = LGA_Avg["Overall_Gap"].sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(10, 7))
Overall.plot(kind="barh", color="grey", ax=ax)
ax.axvline(1, color="red", linestyle="--", label="At standard (1.0)")
ax.invert_yaxis()

ax.set_title("OVERALL RESOURCE GAP BY LGA", fontsize=13, fontweight="bold", color="brown")
ax.set_xlabel("OVERALL GAP (ABOVE 1 = LACKING)", fontweight="bold", color="brown")
ax.set_ylabel("LGA", fontweight="bold", color="brown")
ax.legend()

plt.tight_layout()
plt.show()
```

#### **What the chart shows:**
Each bar shows an overall resource gap for each LGA, which combines teachers, classrooms and toilets into one number.

#### **How the overall gap is calculated:**
Each ratio is divided by its standard (35 for teachers and classrooms, 30 for toilets), and the three results are averaged. Dividing by the standard puts all three resources on the same scale, so they can be combined fairly.

1.0 = exactly at the standard
Above 1.0 = lacking (for example, 2.0 means twice the standard)
Below 1.0 = better than the standard

#### **How to read it:**
The red dashed line marks 1.0. Bars that go past the line belong to LGAs lacking resources overall. The longer the bar, the greater the need.

#### **Insight:**
**Badagry is the most lacking LGA, at 2.00,** which is twice the standards on average. It is followed by **Ojo (1.85), Ajeromi/Ifelodun (1.84), Ibeju/Lekki (1.74) and Alimosho (1.69).**

**Mushin (0.85) and Lagos Mainland (0.97)** are the only LGAs that meet the standards overall. **Kosofe (1.00)** is right at the standard, and **Surulere (1.04) and Oshodi/Isolo (1.06)** are only slightly above it.

Overall, 18 of the 20 LGAs are lacking resources. The most lacking LGAs are mostly growing areas on the edges of Lagos, while the least lacking are older, central areas. These LGAs should be prioritised for support.

```python
# IS THE TEACHER SHORTAGE GETTING WORSE? (2016/17 - 2019/20)
# WHAT TO LOOKS AT -Teacher ratio from 2016 to 2019

Trend = full_df[full_df["Year_Start"] <= 2019]

Trend_Table = Trend.groupby(["Year_Start", "Level"])["Learner_Teacher_Ratio"].mean().unstack()
Trend_Table = Trend_Table[["Primary", "JSS", "SSS"]]

print(Trend_Table.round(1))

fig, ax = plt.subplots(figsize=(10, 6))
Trend_Table.plot(kind="line", marker="o", linewidth=2, ax=ax)
ax.axhline(35, color="red", linestyle="--", label="Standard (35)")

ax.set_xticks([2016, 2017, 2018, 2019])
ax.set_xticklabels(["2016/17", "2017/18", "2018/19", "2019/20"])

ax.set_title("AVERAGE STUDENTS PER TEACHER, 2016/17 TO 2019/20", fontsize=13, fontweight="bold", color="brown")
ax.set_xlabel("SCHOOL YEAR", fontweight="bold", color="brown")
ax.set_ylabel("STUDENTS PER TEACHER", fontweight="bold", color="brown")
ax.legend()

plt.tight_layout()
plt.show()
```

#### **What the chart shows:**
Each line shows the average students per teacher for one school level (Primary, JSS and SSS) across the 20 LGAs, from 2016/17 to 2019/20. Only teacher ratios are shown, because the 2016–2020 report did not record classrooms or toilets.

#### **How to read it:**
The red dashed line marks the standard of 35 students per teacher. A line going upwards means each teacher is handling more students over time, so the shortage is getting worse.

#### **Insight:**
The teacher shortage got worse every year at every school level.

**Primary** rose from about 38 to 53 students per teacher.
**JSS** rose from about 38 to 50 students per teacher.
**SSS** rose from about 25 to 32 students per teacher, still within the standard but moving closer to it.

The biggest jump was between 2017/18 and 2018/19, when the primary ratio rose from about 37 to 49. By 2019/20, both Primary and JSS were well above the standard.

This upward trend shows that student numbers grew faster than teacher numbers. It is the pattern the machine learning model will learn, in order to forecast which LGAs are likely to face worse shortages in future.
