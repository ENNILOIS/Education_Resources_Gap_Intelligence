The raw NMIS dataset has 28 columns. Only the columns needed to compare schools with the UBE minimum standards (35 students per teacher, 35 per classroom, 30 per toilet), to locate them and to identify them were kept.

The following columns were removed:

| Column(s) | Reason for removal |
|---|---|

```python
# Group School Levels

# Make the school type text lower case, so "Primary" and "primary" look the same
school_type = df_clean["facility_type_display"].fillna("").str.lower()

# True/False: Does the school offer each level?
df_clean["Offers_Primary"] = school_type.str.contains("primary") & (school_type != "pre-primary only")
df_clean["Offers_JSS"] = school_type.str.contains("junior")
df_clean["Offers_SSS"] = school_type.str.contains("senior")

# One simple level name for each school
def get_level(row):
    text = str(row["facility_type_display"]).lower()
    p = row["Offers_Primary"]
    j = row["Offers_JSS"]
    s = row["Offers_SSS"]

    if text == "pre-primary only":
        return "Pre-primary"
    if "vocational" in text or "technical" in text:
        return "Technical/Vocational"
    if p and not j and not s:
        return "Primary"
    if j and not p and not s:
        return "JSS"
    if s and not p and not j:
        return "SSS"
    if j and s and not p:
        return "JSS & SSS"
    if p and (j or s):
        return "Unknown"

df_clean["Level"] = df_clean.apply(get_level, axis=1)

# Schools that teach more than one level (their ratios mix levels)
df_clean["Combined_Levels"] = df_clean["Level"].isin(["JSS & SSS", "Primary & Secondary"])

# We no longer need the original column
df_clean = df_clean.drop(columns=["facility_type_display"])

df_clean["Level"].value_counts()
```

```python
import re

# Fix broken characters such as "GovâT"
def fix_characters(text):
    text = str(text)
    try:
        return text.encode("latin-1").decode("utf-8")
    except (UnicodeEncodeError, UnicodeDecodeError):
        return text

# Short forms and their full words
short_forms = {
    r"\bpry\b\.?": "Primary ",
    r"\bpri\b\.?": "Primary ",
    r"\bsch\b\.?": "School ",
    r"\bsnr\b\.?": "Senior ",
    r"\bjnr\b\.?": "Junior ",
    r"\bsec\b\.?": "Secondary ",
    r"\bnur\b\.?": "Nursery ",
    r"\bgrammer\b": "Grammar",
}

def clean_school_name(name):
    name = fix_characters(name)
    for short, full in short_forms.items():
        name = re.sub(short, full, name, flags=re.IGNORECASE)
    name = re.sub(r"\s+", " ", name)      # remove extra spaces
    return name.strip(" .,").title()       # trim and use title case

df_clean["School_Name"] = df_clean["facility_name"].apply(clean_school_name)
df_clean["Ward"] = df_clean["ward"].fillna("").apply(fix_characters).str.strip().str.title()
df_clean["Community"] = df_clean["community"].fillna("").apply(fix_characters).str.strip().str.title()

df_clean = df_clean.drop(columns=["facility_name", "ward", "community"])

df_clean[["School_Name", "Ward", "Community"]].head(10)
```

### Step 5: Cleaning school names, wards and communities

School names in the raw survey were written inconsistently: short forms ("Pry", "Sch", "Snr"), mixed capital letters, extra spaces, spelling errors ("Grammer") and broken characters caused by a text-encoding problem (for example "GovâT").

This matters for three reasons:
1. **Searching:** users look up schools by name and by area, so names, wards and communities must be written consistently to be found.
2. **Duplicates:** the duplicate check in Step 11 compares school name, LGA and ward; inconsistent spellings would hide duplicate records.
3. **Presentation:** clean names make the results readable and professional.

Short forms were expanded to full words, broken characters were repaired, extra spaces were removed, and all names were put in title case.

### Step 9: Checking for outliers

Some ratios in the survey were impossible, for example more than 900 students per teacher or 1 student per classroom. These come from recording errors in the original survey. Left in, they would distort rankings (a school with an impossible low ratio would appear least crowded), averages and charts.

Box plots were used to inspect each ratio. Values outside realistic limits were set to missing and flagged in `Unreliable_Values`:

| Ratio | Lowest allowed | Highest allowed |
|---|---|---|
| Students per teacher | 4 | 250 |
| Students per classroom | 5 | 300 |
| Students per toilet | 5 | 1,000 |


```python
# Add the UBE Labels

# 1. A label for each ratio
def ratio_label(value, standard):
    if pd.isna(value):
        return "Insufficient data – verify with school"
    if value <= standard:
        return "Within UBE benchmark"
    return "Above UBE benchmark – check capacity"

df_clean["Teacher_Label"] = df_clean["Students_Per_Teacher"].apply(ratio_label, standard=35)
df_clean["Classroom_Label"] = df_clean["Students_Per_Classroom"].apply(ratio_label, standard=35)
df_clean["Toilet_Label"] = df_clean["Students_Per_Toilet"].apply(ratio_label, standard=30)

# 2. One overall label per school
def overall_label(row):
    values = [row["Students_Per_Teacher"], row["Students_Per_Classroom"], row["Students_Per_Toilet"]]
    stds = [35, 35, 30]

    if any(pd.isna(v) for v in values):
        return "Insufficient data – verify with school"

    # how many times above the standard, for each ratio that is above it
    times_above = [v / s for v, s in zip(values, stds) if v > s]

    if len(times_above) == 0:
        return "Within UBE benchmark"
    if len(times_above) >= 2 and max(times_above) >= 1.5:
        return "High-pressure facilities – visit before enrolment"
    return "Above UBE benchmark – check capacity"

df_clean["Overall_Label"] = df_clean.apply(overall_label, axis=1)

# 3. Pressure score: 1.0 = exactly at the standards, higher = more crowded
df_clean["Pressure_Score"] = (
    (df_clean["Students_Per_Teacher"] / 35 +
     df_clean["Students_Per_Classroom"] / 35 +
     df_clean["Students_Per_Toilet"] / 30) / 3
).round(2)

# 4. Does the school have all three ratios?
df_clean["Data_Complete"] = df_clean[ratio_cols].notna().all(axis=1)

print(df_clean["Overall_Label"].value_counts())
print("Schools with full data:", df_clean["Data_Complete"].sum())
```


### Step 10: Comparing schools with the UBE standards

Each ratio was compared with its UBE minimum standard (35 students per teacher, 35 per classroom, 30 per toilet) and given a neutral label: "Within UBE benchmark", "Above UBE benchmark – check capacity" or "Insufficient data – verify with school".

An overall label combines the three ratios. Schools above the standard on two or more measures, with at least one 1.5 times above it, are labelled "High-pressure facilities – visit before enrolment".

A pressure score (the average of each ratio divided by its standard) allows schools to be sorted from least to most crowded; a score of 1.0 means exactly at the standards. The labels describe crowding and facilities only, not teaching quality.
